```python
d = 3
noise_model = "bit-flip"
noise = 0.2
```

```python
def create_lattice(d):

    coords = []

    for y in range(2*d - 1):
        for x in range(2*d - 1):
            if x%2 == y%2:
                coords.append((x,y))

    lattice_coords = {coord:i for i,coord in enumerate(coords)}

    return lattice_coords
```

```python
lattice_coords = create_lattice(3)
```

```python
def get_xchecks(d, lattice_coords):

    xchecks = []
    coords = lattice_coords.keys()
    for y in range(2*d - 1):
        for x in range(2*d - 1):
            #Ancillas sit at x%2 != y%2, vertext operators sit at x%2 == 1
            if x%2 == 1 and y%2 == 0:
                    neighbors = [(x+1,y), (x-1,y), (x,y+1), (x,y-1)]
                    xchecks.append([lattice_coords[neighbor] for neighbor in neighbors if neighbor in coords])

    return xchecks

def get_zchecks(d, lattice_coords):
    zchecks = []
    coords = lattice_coords.keys()
    for y in range(2*d - 1):
        for x in range(2*d - 1):
            #Ancillas sit at x%2 != y%2, vertext operators sit at y%2 == 1
            if x%2 == 0 and y%2 == 1:
                    neighbors = [(x+1,y), (x-1,y), (x,y+1), (x,y-1)]
                    zchecks.append([lattice_coords[neighbor] for neighbor in neighbors if neighbor in coords])

    return zchecks
```

```python
x_checks = get_xchecks(d,lattice_coords)
```

```python
z_checks = get_zchecks(d, lattice_coords)
```
